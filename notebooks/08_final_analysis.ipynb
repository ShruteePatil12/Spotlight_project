## Project Title

**Spotlight: Interest-Constrained Recommendation for Emerging Creators**

## 1. Research Overview

Short-video recommendation platforms primarily aim to recommend content that matches users' interests and maximizes engagement. However, recommendation systems that rely heavily on historical interactions and popularity signals may provide greater exposure to already-established creators.

This creates a potential visibility challenge for Emerging creators, whose content may receive fewer opportunities to reach interested audiences.

Spotlight investigates an interest-constrained, fairness-aware recommendation approach that attempts to improve Emerging creator exposure while preserving recommendation relevance.

## 2. Research Motivation

The project is motivated by three objectives:

- Preserve user interest and recommendation relevance.
- Improve exposure opportunities for Emerging creators.
- Explore the trade-off between recommendation quality and creator exposure fairness.

## 3. Research Question

**Can an interest-constrained, fairness-aware recommendation strategy improve exposure opportunities for Emerging creators while maintaining recommendation relevance?**

## 4. Project Approach

The proposed framework combines:

1. User-interest modelling.
2. Baseline collaborative filtering.
3. Creator popularity classification.
4. Controlled exploration.
5. Fairness-aware recommendation ranking.
6. Offline evaluation using temporal holdout data.

## 5. Research Dataset

The project uses the KuaiRec dataset, containing user-video interaction records and video metadata.

The experiments focus on 1,411 users and 3,327 videos from the small interaction matrix.

## 6. Dataset Description

### 6.1 Dataset Source

The experiments use the **KuaiRec dataset**, a real-world short-video recommendation dataset containing user-video interactions and video metadata.

The primary interaction data used is `small_matrix.csv`.

### 6.2 Dataset Characteristics

| Attribute | Value |
|---|---:|
| Users | 1,411 |
| Videos | 3,327 |
| Original interactions | 4,676,570 |
| Interactions after date cleaning | 4,494,578 |

### 6.3 Data Preprocessing

The following preprocessing operations were performed:

1. Loaded user-video interaction records.
2. Removed records with missing date values.
3. Converted date fields into appropriate datetime formats.
4. Processed watch ratios and capped extreme values at 3.0 to create a bounded `watch_score`.
5. Connected video metadata with creator information.
6. Extracted detailed video tags for interest modelling.
7. Aggregated creator popularity using their total `show_cnt`.

### 6.4 Temporal Train-Test Split

A chronological split was used to preserve the time ordering of interactions.

| Dataset | Period | Interactions |
|---|---|---:|
| Training | July 5 – August 29, 2020 | 4,387,382 |
| Testing | August 30 – September 5, 2020 | 107,196 |

The training period was used to construct recommendation models, while the later test period was used for offline evaluation.

### 6.5 Creator Classification

Creators were classified into two groups using their aggregated popularity within the experimental creator population.

- **Emerging creators:** Bottom 25% based on total popularity.
- **Established creators:** Remaining 75%.

The resulting population consisted of:

| Creator Group | Number of Creators | Population Share |
|---|---:|---:|
| Emerging | 508 | Approximately 25% |
| Established | 1,523 | Approximately 75% |
| Total | 2,031 | 100% |

This classification provides the basis for evaluating creator exposure distribution.

## 7. Proposed Methodology: Spotlight Framework

Spotlight combines user-interest modelling with controlled, fairness-aware exploration to improve exposure opportunities for Emerging creators.

### 7.1 Baseline Recommendation Model

A user-based collaborative filtering approach was implemented using cosine similarity.

The model identifies users with similar interaction patterns and uses their preferences to generate recommendations.

This provides the baseline recommendation strategy against which the proposed approaches are evaluated.

### 7.2 User-Interest Modelling

User interests were derived from historical video interactions and detailed video tags.

For each user-tag combination, interaction frequency and average watch score were calculated.

A smoothed interest score was used to reduce the influence of tags with very few interactions.

The smoothing parameter was set to 5.

User interests were categorized into:

- **Established interests:** At least 8 interactions with a tag.
- **Weak interests:** Between 1 and 7 interactions with a tag.
- **Unseen interests:** No previous interactions.

### 7.3 Interest-Constrained Exploration

The exploration strategy identifies videos associated with users' weak interests.

Instead of recommending completely unrelated content, the framework explores content connected to interests that users have interacted with less frequently.

Different exploration allocations were evaluated:

- 0% exploration
- 10% exploration
- 20% exploration
- 30% exploration

The remaining recommendation slots were filled using normal interest-based recommendations.

### 7.4 Fairness-Aware Ranking

Creator popularity information was incorporated into the exploration ranking process.

Emerging creators received an additional fairness boost of 0.08.

The combined exploration score was calculated as:

**Combined Exploration Score = Smoothed Interest Score + Fairness Boost**

The fairness boost was applied to Emerging creators, while Established creators received no additional boost.

This allows creator exposure considerations to influence exploration ranking while retaining user-interest relevance as a scoring component.

### 7.5 Final Recommendation Generation

The final fairness-aware recommendation list contained:

- 7 normal interest-based recommendations.
- 3 fairness-aware exploration recommendations.

Each user received exactly 10 unique video recommendations.

### 7.6 Framework Objective

The overall objective of Spotlight is to investigate whether creator fairness can be incorporated into interest-constrained recommendation ranking without substantially compromising recommendation relevance.

## 8. Experimental Setup

### 8.1 Evaluation Strategies

Three recommendation strategies were compared:

**1. Baseline Recommendation (0% Exploration)**

Uses interest-based recommendations without an exploration allocation or additional creator fairness boost.

**2. Ordinary Exploration (30%)**

Allocates 30% of recommendation slots to weak-interest exploration and the remaining 70% to normal interest-based recommendations.

**3. Fairness-Aware Exploration (30%)**

Allocates 30% of recommendation slots to fairness-aware exploration, using a creator fairness boost of 0.08. The remaining 70% consists of normal interest-based recommendations.

### 8.2 Recommendation Configuration

| Parameter | Configuration |
|---|---:|
| Total users | 1,411 |
| Recommendations per user | 10 |
| Total recommendations per strategy | 14,110 |
| Normal recommendation slots | 7 |
| Exploration recommendation slots | 3 |
| Fairness boost | 0.08 |

### 8.3 Evaluation Metrics

The following metrics were used to evaluate recommendation quality and creator fairness.

**HitRate@10**

Measures the proportion of users who receive at least one recommendation matching their interactions in the held-out test period.

**Recommendation Slot Hit Rate**

Measures the proportion of all recommended video slots that match held-out test interactions.

**Creator Exposure Share**

Measures the percentage of recommendation slots allocated to Emerging and Established creators.

**Creator Coverage**

Measures the percentage of creators in a group who receive at least one recommendation.

**Creator Diversity**

Measures the number of distinct creators represented in each user's Top-10 recommendation list.

**Emerging Creator Exposure Gap**

Measures the difference between Emerging creator exposure and their population share.

A value closer to zero indicates exposure closer to population representation.

### 8.4 Offline Evaluation

The models were evaluated using interactions from the held-out chronological test period.

For each strategy, 10 unique videos were recommended to every user.

Duplicate user-video recommendations were removed during recommendation generation to ensure that each recommendation list contained 10 distinct videos.

The evaluation examines both recommendation relevance and creator exposure outcomes.

## 9. Experimental Results and Comparative Analysis

The three recommendation strategies were evaluated using the same users, test interactions, and recommendation list size.

### 9.1 Recommendation Quality

| Strategy | HitRate@10 (%) | Slot Hit Rate (%) |
|---|---:|---:|
| Baseline (0%) | 89.30 | 22.71 |
| Ordinary Exploration (30%) | 88.94 | 21.79 |
| Fairness-Aware (30%) | 88.09 | 21.35 |

**Observation:**

The fairness-aware strategy achieved a HitRate@10 of 88.09%, compared with 89.30% for the baseline.

This represents an absolute reduction of 1.21 percentage points, indicating a modest relevance trade-off.

### 9.2 Emerging Creator Exposure

| Strategy | Emerging Exposure (%) | Established Exposure (%) |
|---|---:|---:|
| Baseline | 10.08 | 89.92 |
| Ordinary Exploration | 10.26 | 89.74 |
| Fairness-Aware | 12.67 | 87.33 |

**Observation:**

The fairness-aware strategy increased Emerging creator exposure by 2.59 percentage points compared with baseline.

Ordinary exploration produced only a 0.18 percentage-point improvement.

### 9.3 Emerging Creator Exposure Gap

The Emerging creator population share was approximately 25.01%.

| Strategy | Exposure Gap (pp) |
|---|---:|
| Baseline | -14.93 |
| Ordinary Exploration | -14.76 |
| Fairness-Aware | -12.34 |

**Observation:**

The fairness-aware strategy reduced the Emerging creator exposure gap by 2.59 percentage points compared with baseline.

However, the remaining negative gap indicates that Emerging creators still receive less exposure than their population representation.

### 9.4 Creator Coverage

| Strategy | Emerging Coverage (%) | Established Coverage (%) |
|---|---:|---:|
| Baseline | 47.24 | 54.43 |
| Ordinary Exploration | 46.46 | 53.51 |
| Fairness-Aware | 48.03 | 53.51 |

**Observation:**

Emerging creator coverage increased from 47.24% under baseline to 48.03% under fairness-aware exploration.

This indicates a modest increase in the number of distinct Emerging creators receiving recommendations.

### 9.5 Creator Diversity

| Strategy | Mean Unique Creators | Median |
|---|---:|---:|
| Baseline | 9.84 | 10 |
| Ordinary Exploration | 9.86 | 10 |
| Fairness-Aware | 9.87 | 10 |

**Observation:**

Fairness-aware exploration achieved the highest mean creator diversity, although the improvement over baseline was small.

### 9.6 Overall Comparative Analysis

The experimental findings demonstrate that:

1. Ordinary exploration alone produces limited improvement in Emerging creator exposure.
2. Incorporating a creator fairness boost increases Emerging creator exposure.
3. Fairness-aware recommendations retain a relatively high HitRate@10.
4. Creator coverage and diversity improve modestly under the fairness-aware strategy.
5. A measurable relevance–fairness trade-off remains.

Overall, the results support further investigation of interest-constrained, fairness-aware exploration as an approach to improving creator exposure opportunities.v

## 10. Research Contributions

The Spotlight framework contributes an experimental approach to balancing user-interest relevance and creator exposure fairness in short-video recommendation systems.

### 10.1 Interest-Constrained Exploration

Rather than exploring completely unrelated content, Spotlight uses users' weak interests to identify relevant exploration opportunities.

This provides a way to introduce less-frequently explored content while retaining a connection to historical user preferences.

### 10.2 Creator-Aware Recommendation Ranking

The framework incorporates creator popularity groups into exploration ranking.

An additional fairness boost is assigned to Emerging creators, allowing creator exposure considerations to influence recommendation selection.

### 10.3 Exposure-Based Fairness Evaluation

The project evaluates creator fairness through multiple dimensions:

- Recommendation exposure share.
- Creator coverage.
- Emerging creator exposure gap.
- Creator diversity.

This provides a broader evaluation than recommendation relevance alone.

### 10.4 Relevance–Fairness Analysis

The experiments demonstrate that fairness-aware exploration can increase Emerging creator exposure while retaining relatively high HitRate@10.

The results also highlight the trade-off between recommendation relevance and creator exposure fairness.

## 11. Key Research Findings

The major findings are:

1. Emerging creators represent approximately 25.01% of the experimental creator population but receive only 10.08% of baseline recommendation exposure.
2. Ordinary exploration increases Emerging creator exposure only slightly, reaching 10.26%.
3. Fairness-aware exploration increases Emerging creator exposure to 12.67%.
4. The Emerging creator exposure gap decreases from -14.93 to -12.34 percentage points.
5. HitRate@10 decreases from 89.30% under baseline to 88.09% under fairness-aware exploration.
6. Mean creator diversity increases from 9.84 to 9.87 unique creators per user.

Together, these findings indicate that incorporating creator fairness into interest-constrained exploration can improve Emerging creator exposure, although the resulting exposure remains below population representation.

## 12. Research Significance

The Spotlight framework investigates creator exposure as an additional objective in recommendation system design.

It demonstrates the potential of combining personalized relevance with creator-aware exploration instead of relying exclusively on historical interaction patterns.

The experimental findings provide a foundation for further research into recommendation strategies that consider both user interests and opportunities for less-established creators.
